------------------------------
## 造成行數差距的 3 大關鍵原因
## 1. 前置宣告與環境設定 (C++ 多 6-7 行)

* Python：開箱即用，不需要引進任何模組。
* C++：必須引入標頭檔（如 #include <iostream>、#include <vector>）、宣告命名空間 using namespace std; 以及建立主函式 int main()。

## 2. 資料輸入與陣列初始化 (C++ 多 5-6 行)

* Python：一行 list(map(int, input().split())) 就能同時完成「讀取一行、空白切割、轉成整數、包成清單」四個動作。
* C++：需要先宣告固定長度的 vector，再利用 for 迴圈跑三次 cin 逐一讀入數字。

## 3. 語法結構與大括號 (C++ 多 8-10 行)

* Python：使用縮進（Indentation）區隔程式區塊，且支援單行多變數同時賦值（如 a, b, c = sides[0], sides[1], sides[2]）。
* C++：每個條件判斷與迴圈都需要 {} 大括號換行，且變數必須逐行單獨宣告型態。
