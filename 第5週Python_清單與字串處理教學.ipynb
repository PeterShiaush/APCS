### **1. [教學區]：map 與 split 的大合體**

在 APCS 實作題中，測資通常是「一行好幾個數字」。這行程式碼你一定要背起來，它是 Python 解題的生命線：

`nums = list(map(int, input().split()))`

* `input()`：讀入整行字串 `"10 20 30"`。
* `.split()`：把字串切開變成 `["10", "20", "30"]`。
* `map(int, ...)`：map() 會像工廠的傳送帶一樣，把剛才切好的清單 ["10", "20", "30"] 裡的每一個元素，**逐一**丟進 **int()** 函數裡進行轉換。過程：　"10" → 10，　"20" → 20，　"30" → 30。

* `list(...)`：最後包裝成一個 整數的清單 `[10, 20, 30]`。

### **題目 2：[a015: 矩陣翻轉](https://zerojudge.tw/ShowProblem?problemid=a015) (回家，自我挑戰，APCS將由 2級 開始跳到 3級)**

* **目標**：將橫的變直的。
* **提示**：這題會用到「巢狀迴圈（迴圈裡面還有迴圈）」。

image.png

```
import sys # 引入 sys 模組，讓我們可以從標準輸入 (sys.stdin) 讀取資料，這在像 ZeroJudge 這種線上測驗平台很常用。

for line in sys.stdin: # 這個迴圈會一行一行地讀取你輸入的資料。當資料讀完，迴圈就會結束。
    # 讀取每組測試資料的第一行，通常包含矩陣的『列數』(row) 和『行數』(col)
    # line.strip().split()：先移除前後空白，再用空白把數字字串切開。
    # map(int, ...)：把切開的字串轉換成整數。
    row, col = map(int, line.strip().split()) # Row=2 (列)：水平方向（橫排）的資料，Column=3 (欄/行)：垂直方向（直行，柱子）的資料
    matrix = [] # 建立一個空的清單 (list)，用來存放整個矩陣（想像成一個大表格）。

    # 讀取矩陣的每一列資料
    for i in range(row): # 這個迴圈會執行 row 次，每次讀取矩陣的一列。
        # input().split()：讀取一行數字字串，然後用空白切開。
        # map(int, ...)：把切開的字串轉換成整數。
        # list(...)：把轉換後的整數集合變成一個清單。
        # matrix.append(...)：把這一列的數字清單加到 matrix 大清單中。
        matrix.append(list(map(int, input().split())))

    # 翻轉輸出邏輯：將原本的矩陣 (row x col) 翻轉成 (col x row) 的形式。
    # 想像把表格橫著看變成直著看，原本的行變成新的列，原本的列變成新的行。
    for j in range(col): # 外層迴圈：控制翻轉後矩陣的『列』，也就是原本矩陣的『行』。
        for i in range(row): # 內層迴圈：控制翻轉後矩陣的『行』，也就是原本矩陣的『列』。
            # matrix[i][j]：這裡的 i 和 j 順序和我們讀取時的 matrix[列][行] 是相反的！
            # 這就是實現翻轉的關鍵：把原本 (i列, j行) 的元素，印到 (j新列, i新行) 的位置。
            # end=" "：表示每次印出數字後，不要換行，而是用一個空格隔開，讓數字印在同一行。
            print(matrix[i][j], end=" ")
        print() # 當內層迴圈結束（印完翻轉後的一整列），呼叫這個 print() 函式，會產生一個換行，讓下一列從新的行開始印出。
```
