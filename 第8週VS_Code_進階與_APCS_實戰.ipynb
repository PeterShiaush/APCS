### **1. [教學區]：別再只用 print 找 Bug！**

**VS Code** 專業開發者會使用 **偵錯器 (Debugger)** 來觀察程式執行的過程。

* **斷點 (Breakpoint)**：在行號左邊，點一下**紅點** 🔴，程式執行到這裡會「**暫停**」。
* **變數監看**：在暫停時，可以直接看到所有變數目前的數值。
* **快捷鍵**：
* `F5`：開始偵錯。
* `F10`：跳到下一步 (Step Over)。



#### 1.1 **[教學範例區]：VS Code 簡易偵錯範例**

假設我們寫了一個計算「數字總和」的程式，但它好像有時候會算錯。我們來用偵錯器找出問題！

#### New File: **test00.py**

image.png

```
def calculate_sum(numbers):
    total = 0
    # 故意將迴圈範圍減少 1，讓它少加最後一個數字
    for i in range(len(numbers) - 1): # <-- 這裡有個故意的 Bug！
        print(f"目前 i = {i}, 正在加的數字是 {numbers[i]}")
        total += numbers[i]
    return total

# 測試我們的函數
my_list = [10, 20, 30, 40, 50]
expected_sum = 150 # 10+20+30+40+50 = 150

actual_sum = calculate_sum(my_list)

print(f"我的列表是：{my_list}")
print(f"預期總和：{expected_sum}")
print(f"實際計算結果：{actual_sum}")

if actual_sum != expected_sum:
    print("糟糕！計算結果不符，有 Bug 喔！")
else:
    print("恭喜！計算正確！")
```

#### 1.2 **如何使用偵錯器 (Debugger) 找出錯誤**

1.  **設定斷點 (Breakpoint)**：
    *   請在上面 `total += numbers[i]` 這行的**行號左邊**點一下，你會看到一個紅點 。下圖，🔴 ，就是「斷點」。
    *   image.png
    
    *   這告訴程式：執行到這裡時請「暫停」下來。

2.  **開始偵錯 (F5)**： 會彈出 "**Python Debugger**"、**"Python File**" 等選項供您點選。
    *   按下鍵盤上的 `F5`（或點選你的編輯器裡的偵錯按鈕），程式就會開始執行。
    *   當程式跑到你設定的斷點時，它會停下來。

3.  **觀察變數**：
    *   當程式暫停時，你會看到**編輯器左邊**，有一個「**變數**」(**VARIABLES**)或「監看式」的視窗。
    *   仔細觀察 `total` 和 `i` 這兩個變數的數值，它們會隨著每次迴圈而變化。
    *   你可能會發現 `i` 的範圍或 `numbers[i]` 的取值有些不對勁，導致 `total` 的結果錯誤。


image.png



4.  **逐步執行 (F10)**：
    *   你可以按下 `F10`，程式就會「一步一步」地往下執行。
    *   每次按 `F10`，你都可以觀察變數的變化，直到你找到是哪一步出了問題。

現在，你會發現這個範例中確實有一個 Bug！透過這個過程，你已經學會了如何設定斷點、觀察變數以及逐步執行程式，並成功找出並理解問題所在。這就是偵錯器的魔法！
