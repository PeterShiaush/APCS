**語法結構：**

```
for i in range(??):
    # 重複執行的程式碼 (記得縮排！) print()， 內含 "迴圈第", i, "次"
```

* `range(5)`：產生 `0, 1, 2, 3, 4` (從 0 開始，**不包含** 5)。
