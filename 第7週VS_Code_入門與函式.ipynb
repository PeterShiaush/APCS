### **1. [教學區]：為什麼要從 Colab 搬家到 VS Code？**

雖然 Colab 方便，但專業工程師使用 **VS Code (Visual Studio Code)** 是因為：

* **執行速度**：直接使用電腦效能，不需等待雲端分配機器。

### **1. [教學區]：什麼是函式 (Function)？**

函式就像是一台「自動販賣機」。你投入參數（投幣），它執行內部的邏輯（選擇飲料），最後傳回 (return) 結果（掉出飲料）。

* **關鍵字**：`def` (定義函式)。
* **縮排**：函式內部的程式碼必須向內縮排。
* **傳回值**：使用 `return` 將結果送出，這跟 `print` 不同，`return` 的結果可以被存在變數中。
